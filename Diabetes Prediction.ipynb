#Loading modules and dataset

```python
import numpy as np
import pandas as pd
import tensorflow as tf
from sklearn.preprocessing import StandardScaler
from imblearn.over_sampling import RandomOverSampler
```

```python
df = pd.read_csv('diabetes_data_upload.csv')
```

#Preprocessing the dataset


```python
df["Gender"] = df["Gender"].map({"Male": 1, "Female": 0})
df["class"] = df["class"].map({"Positive": 1, "Negative": 0})

binary_cols = [
    "Polyuria", "Polydipsia", "sudden weight loss", "weakness", "Polyphagia",
    "Genital thrush", "visual blurring", "Itching", "Irritability",
    "delayed healing", "partial paresis", "muscle stiffness", "Alopecia", "Obesity"
]

for col in binary_cols:
    df[col] = df[col].map({"Yes": 1, "No": 0})

df.head()
```

```python
def scale_dataset(dataset, over_sample=False):
  x = dataset.drop(columns=['class'])
  y = dataset['class']

  scaler = StandardScaler()
  x = scaler.fit_transform(x)

  if over_sample:
    ros = RandomOverSampler()
    x, y = ros.fit_resample(x, y)

  return x, y
```

```python
from sklearn.model_selection import train_test_split
x, y = scale_dataset(df, over_sample=True)
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2)
```

#Create and Train the model


```python
model = tf.keras.Sequential([
    tf.keras.layers.Dense(32, activation='relu'),
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.Dense(16, activation='relu'),
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.Dense(1, activation='sigmoid')
])
```

```python
model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
              loss=tf.keras.losses.BinaryCrossentropy(),
              metrics=['accuracy'])
```

```python
history = model.fit(x_train, y_train, epochs=50, batch_size=32, validation_split=0.2, verbose=0)
```

#Evaluate the model


```python
model.evaluate(x_test, y_test)
```

### Clinical Symptoms Glossary
#### This cell was made with AI

Here is a description of the symptoms used as inputs for this diabetes risk assessment model:

* **Polyuria**: Excessive or abnormally large passage of urine. This is a classic primary symptom of diabetes.
* **Polydipsia**: Excessive thirst or drinking, often triggered by dehydration caused by polyuria.
* **Sudden Weight Loss**: Unexplained and rapid loss of weight, which occurs because the body cannot properly utilize glucose for energy and begins burning fat and muscle instead.
* **Weakness/Fatigue**: A persistent feeling of tiredness or exhaustion, stemming from the body's cells being deprived of glucose.
* **Polyphagia**: Excessive hunger or increased appetite, driven by the body's inability to feed glucose into cellular metabolism.
* **Genital Thrush**: A yeast infection common in people with elevated blood glucose levels, as yeast thrives in sugar-rich environments.
* **Visual Blurring**: Temporary blurry vision caused by high blood sugar pulling fluid from the lenses of the eyes, affecting the ability to focus.
* **Itching**: Pruritus or generalized itchy skin, often caused by dry skin or poor circulation associated with high blood sugar.
* **Irritability**: Fluctuations in mood or heightened irritability, which can be directly linked to rapidly fluctuating blood sugar levels.
* **Delayed Healing**: Slow healing of cuts, sores, or infections, occurring due to poor circulation and diminished immune function caused by prolonged high blood glucose.
* **Partial Paresis**: Mild or partial muscle weakness/paralysis, which can be an early sign of diabetic neuropathy (nerve damage).
* **Muscle Stiffness**: Muscle rigidity or tightness, sometimes related to electrolyte imbalances or joint stiffness in diabetes.
* **Alopecia**: Sudden hair loss or thinning, which can be related to the chronic stress or underlying autoimmune factors associated with diabetes.
* **Obesity**: A high body mass index (BMI) or excessive body fat, which is a major risk factor for developing insulin resistance and Type 2 diabetes.
