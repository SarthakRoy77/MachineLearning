#Importing all dependencies


```python
import pandas as pd
import numpy as np
from imblearn.over_sampling import RandomOverSampler
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
```

#Accessing, Scaling and Spliting Data

```python
dataset=pd.read_csv("wine.data")
```

```python
dataset.head()
```

```python
def scale_dataset(dataset, oversample=False):
  """Function to scale and return x and y with option to oversample"""
  x = dataset.drop(columns="class")
  y = dataset["class"]

  scaler = StandardScaler()
  x = scaler.fit_transform(x)

  if oversample:
    ros = RandomOverSampler()
    x, y = ros.fit_resample(x, y)

  return x, y
```

```python
train_dataset, test_dataset = np.split(dataset.sample(frac=1), [int(0.8 * len(dataset))])

x_train, y_train = scale_dataset(train_dataset, oversample=True)
x_test, y_test = scale_dataset(test_dataset, oversample=False)
```

#Fitting and Predicting

```python
lg_model = LogisticRegression()
lg_model.fit(x_train, y_train)
y_pred = lg_model.predict(x_test)
print(classification_report(y_test, y_pred))
```

Model gives 97 percent accuracy
